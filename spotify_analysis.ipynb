```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# ==========================================
# 1. DATASET SELECTION AND READING
# ==========================================
print("--- Step 1: Loading Dataset ---")
# Reading the generated Spotify CSV dataset
df = pd.read_csv('spotify_tracks_data.csv')
print("Dataset successfully loaded into Pandas DataFrame!\n")

# ==========================================
# 2. DATA UNDERSTANDING
# ==========================================
print("--- Step 2: Data Understanding ---")
print(f"Dataset Shape: {df.shape[0]} rows, {df.shape[1]} columns")
print("\nDataset Info:")
print(df.info())
print("\nMissing Values Count before cleaning:")
print(df.isnull().sum())
print("\nFirst 5 Records:")
print(df.head(), "\n")

# ==========================================
# 3. DATA TRANSFORMATION
# ==========================================
print("--- Step 3: Data Transformation ---")
# Task A: Transform column data type if needed (Ensure Year behaves cleanly as categorical group)
df['Release_Year'] = df['Release_Year'].astype(int)

# Task B: Add a new calculated column (Convert Duration from milliseconds to Minutes)
df['Duration_Mins'] = round(df['Duration_MS'] / 60000, 2)

# Task C: Drop unnecessary column to optimize memory space
df.drop(columns=['Duration_MS'], inplace=True)
print("Data transformation complete. Added 'Duration_Mins' and removed raw 'Duration_MS'.\n")

# ==========================================
# 4. HANDLING NULL VALUES (MEAN IMPUTATION)
# ==========================================
print("--- Step 4: Handling Null Values via Mean Imputation ---")
# Calculating means for columns with missing values
streams_mean = df['Streams'].mean()
popularity_mean = df['Popularity'].mean()

print(f"Calculated Mean for Streams: {streams_mean:.2f}")
print(f"Calculated Mean for Popularity: {popularity_mean:.2f}")

# Applying Mean Imputation
df['Streams'] = df['Streams'].fillna(streams_mean)
df['Popularity'] = df['Popularity'].fillna(popularity_mean)

print("Verification - Missing Values Count after Imputation:")
print(df.isnull().sum(), "\n")

# ==========================================
# 5. INSIGHT GENERATION
# ==========================================
print("--- Step 5: Insight Generation ---")
# Insight 1: Total Streams generated by each musical genre
genre_streams = df.groupby('Genre')['Streams'].sum().sort_values(ascending=False).reset_index()
print("\nInsight 1: Total Streams by Genre:")
print(genre_streams)

# Insight 2: Evolutionary change of average track popularity over release years
yearly_popularity = df.groupby('Release_Year')['Popularity'].mean().reset_index()
print("\nInsight 2: Average Track Popularity over Time (Years):")
print(yearly_popularity)

# ==========================================
# 6. DATA VISUALIZATION
# ==========================================
print("\n--- Step 6: Data Visualization ---")
sns.set_theme(style="whitegrid")

# Visualization 1: Total Streams by Music Genre
plt.figure(figsize=(10, 5))
sns.barplot(data=genre_streams, x='Genre', y='Streams', palette='viridis')
plt.title('Total Streams Distribution by Musical Genre', fontsize=14, fontweight='bold')
plt.xlabel('Genre', fontsize=12)
plt.ylabel('Total Streams (Billions)', fontsize=12)
plt.xticks(rotation=15)
plt.tight_layout()
plt.savefig('genre_streams_insight.png', dpi=300)
plt.close()
print("Saved: 'genre_streams_insight.png'")

# Visualization 2: Trendline of Track Popularity over Time
plt.figure(figsize=(10, 5))
sns.lineplot(data=yearly_popularity, x='Release_Year', y='Popularity', marker='o', color='crimson', linewidth=2.5)
plt.title('Average Track Popularity Trends over Time (2018 - 2025)', fontsize=14, fontweight='bold')
plt.xlabel('Release Year', fontsize=12)
plt.ylabel('Average Popularity Score', fontsize=12)
plt.tight_layout()
plt.savefig('yearly_popularity_trend.png', dpi=300)
plt.close()
print("Saved: 'yearly_popularity_trend.png'")

print("\nAll project milestones completed successfully! Ready for code repository export.")
```
